```python
area_w = 4
area_h = 3

def inArea(x, y):
    if((x>=0 and x<=2 and y>=0 and y<=3) or (x>=2 and x<=4 and y>=1 and y<=2)):
        return 1
    else:
        return 0
```

```python
import random as rd
import numpy as np

x = np.array([rd.random()*area_w for _ in range(3000)])
y = np.array([rd.random()*area_h for _ in range(3000)])

vfunc = np.vectorize(inArea)

results = vfunc(x, y)
print(results)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(x[results == 0], y[results == 0], color='orange', alpha=0.3, label='Outside Area')
plt.scatter(x[results == 1], y[results == 1], color='green', alpha=0.5, label='Inside Area')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot with Highlighted Areas')
plt.legend()
plt.grid()
plt.show()
```

```python
S = np.sum(results)*area_w*area_h/len(results)
print(f'Result using Monte Carlo: {S}')
print(f'Result using Geometry:    {2*3+2*1}')
```

Algorithm of detecting if the point is inside the area:
1. Deifine the function of area: 'inArea' with following
- x,y ∈ [0,2],[0,3] || x,y ∈ [2,4],[1,2]
2. Define random points inside the 3*4 area
3. Apply points to function of area to define if the point inside the area
